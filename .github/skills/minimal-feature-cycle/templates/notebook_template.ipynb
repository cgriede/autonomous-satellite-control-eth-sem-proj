```python
# file handling setup do not delete
import sys, os
from pathlib import Path


def change_cwd_root_notebook(root_name: str = "backend", iter: int = 3):
    # __vsc_ipynb_file__ is set by VS Code/Cursor to the notebook's actual path
    notebook_dir = Path(os.path.abspath(__vsc_ipynb_file__)).parent  # type: ignore

    next_parent = notebook_dir.parent
    parent_found = False
    while (parent_found == False):
        if next_parent.name == root_name:
            os.chdir(next_parent)
            sys.path.insert(0, str(next_parent))
            print(f"changed cwd to {str(next_parent)}")
            break
        if iter == 0:
            print(f"WARN; could not locate root \nname: {root_name}, iter {iter} ")
            break
        iter -= 1
        next_parent = next_parent.parent


change_cwd_root_notebook()
_repo_root = Path.cwd().parent
if str(_repo_root) not in sys.path:
    sys.path.insert(0, str(_repo_root))
```
